This tutorial explains [Discrete-time Markov chain](https://en.wikipedia.org/wiki/Discrete-time_Markov_chain#Stationary_distributions).

This is our transition matrix between rain and dry states.

| **Next State** $P_t$ | Dry | Rain |
|------------------------|------|-----|
| **Initial State** $P_{t-1}$ **Dry** | 0.8 | 0.2 |
| **Initial State** $P_{t-1}$ **Rain**  | 0.6 | 0.4 |

```python
import numpy as np

# Transition matrix
P = np.array([[0.8, 0.2],
              [0.6, 0.4]])

# Initial state probabilities (this can be obtained from historical observation)
P0 = np.array([0.875, 0.125])  # [Dry, Rain]
```

### (1) What is the probability that there is rain on the third day?  

$P_0 P^2 = [0.875 \quad 0.125] \begin{bmatrix} 0.8 & 0.2 \\ 0.6 & 0.4 \end{bmatrix}^2$

```python
# Calculate probability on the 3rd day (P0 * P^2)
P2 = np.linalg.matrix_power(P, 2)
P_third_day = np.dot(P0, P2)

print("Probability on the 3rd day:")
print(f"Dry: {P_third_day[0]:.4f}, Rain: {P_third_day[1]:.4f}")
```

### (2) What is the probability that there is rain on the nth day, where n is large?

The probability that there is rain on the $n$ th day is

$P_0 P^n = [0.875 \quad 0.125] \begin{bmatrix} 0.8 & 0.2 \\ 0.6 & 0.4 \end{bmatrix}^n$


And, if $n$ becomes large:

$lim_{n \to \infty} P^n =
\begin{bmatrix}
0.75 & 0.25 \\
0.75 & 0.25
\end{bmatrix}
$

So, the probability that there is rain on the $n$ th day is

$lim_{n \to \infty} P_0 P^n =
[0.875 \quad 0.125]\begin{bmatrix}
0.75 & 0.25 \\
0.75 & 0.25
\end{bmatrix}
$

```python
# Approach 1 (Direct Computation)
print('Approach 1')
P_inf = np.linalg.matrix_power(P, 3)
print("\nProbability on the nth day (n -> infinity):")
print(f"Dry: {P_inf[0, 0]:.7f}, Rain: {P_inf[0, 1]:.7f}")

# Approach 2 (Eigenvector Method)
# Steady-state probabilities as n -> infinity
eigenvalues, eigenvectors = np.linalg.eig(P.T)
steady_state = eigenvectors[:, np.isclose(eigenvalues, 1)].flatten().real
steady_state /= steady_state.sum()  # Normalize
print('')
print('Approach 2')
print("\nSteady-state probabilities (n -> infinity):")
print(f"Dry: {steady_state[0]:.7f}, Rain: {steady_state[1]:.7f}")
```

# Stationary Distribution of a Markov Chain

In a Markov chain, the **stationary distribution** represents the long-run probabilities of being in each state. It is the probability distribution that remains unchanged after applying the transition matrix. 

Let the stationary distribution be

$
\pi = [\pi_1, \pi_2].
$

A stationary distribution $\pi$ satisfies:

$\pi P = \pi, \qquad \pi_1 + \pi_2 = 1.$

For the transition matrix

$
P = \begin{bmatrix}
0.8 & 0.2 \\
0.6 & 0.4
\end{bmatrix},
$

Expanding the fixed-point equation:

$
\pi_1 = 0.8\pi_1 + 0.6\pi_2,
$
$
\pi_2 = 0.2\pi_1 + 0.4\pi_2.
$

Using the condition $\pi_2 = 1 - \pi_1$:

$
\pi_1 = 0.8\pi_1 + 0.6(1 - \pi_1),
$
$
0.8\pi_1 = 0.6,
$
$
\pi_1 = 0.75.
$

Thus,

$
\pi_2 = 0.25.
$

Final Stationary Distribution

$
\boxed{\pi = [0.75,\, 0.25]}
$

As $n \to \infty$, the Markov chain converges to this distribution regardless of the initial state.
